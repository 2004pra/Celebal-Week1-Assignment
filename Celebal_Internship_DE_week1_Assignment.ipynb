# Handling missing Values
My Approach -> As i can see that the there are few text based missing values and numeric based also , this approach mainly focus on not reducing the size of the dataset

1. Text Based missing Columns are :
(i)    what_customers_said
(ii)   seller_name
(iii)  videos
(iv)   seller_information
(v)    variations



2. Numeric Missing Columns :
(i) Discount

```python
# Handling Missing values text and numeric seperately

# Text based are replaced using the word "Not Available" because those
# informations are not available

text_columns = [
    'what_customers_said',
    'seller_name',
    'videos',
    'seller_information',
    'variations'
]

for col in text_columns:
    df[col].fillna("Not Available", inplace=True)


# For Numeric Missing values are replaced using 0 Means No discount for those products

df['discount'].fillna(0, inplace=True)


# Now Lets verify the dataset again

df.isnull().sum()
```

Removing the duplicate Values

```python
df.duplicated().sum()

# this will remove duplicate only if there are more than 0 duplicates
df.drop_duplicates(inplace=True)
```
